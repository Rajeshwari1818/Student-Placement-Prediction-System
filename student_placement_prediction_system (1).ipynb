```python
categorical_features = ['gender','branch','college_tier','volunteer_experience']
```

```python
X_train = pd.get_dummies(X_train, columns=categorical_features)
X_test = pd.get_dummies(X_test, columns=categorical_features )
```

TASK-10(Train the Machine Learning Model)

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
```

```python
logistic_model = LogisticRegression(max_iter=1000)
random_forest_model = RandomForestClassifier(n_estimators=100, random_state=42)
knn_model = KNeighborsClassifier()
svm_model = SVC()
```

```python
logistic_model.fit(X_train, y_train)
random_forest_model.fit(X_train, y_train)
knn_model.fit(X_train, y_train)
svm_model.fit(X_train, y_train)
```

```python
y_pred_logistic = logistic_model.predict(X_test)
y_pred_random_forest = random_forest_model.predict(X_test)
y_pred_knn = knn_model.predict(X_test)
y_pred_svm = svm_model.predict(X_test)
```

I trained four models: Logistic Regression, Random Forest, SVM, and KNN.

Observation: Training different models allowed me to compare their performance.

TASK=11(Model Evaluation)

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
models = {
    'Logistic Regression': y_pred_logistic,
    'Random Forest': y_pred_random_forest,
    'SVM': y_pred_svm,
    'KNN': y_pred_knn
}

results = []

for name, predictions in models.items():
    results.append([
        name,
        accuracy_score(y_test, predictions),
        precision_score(y_test, predictions),
        recall_score(y_test, predictions),
        f1_score(y_test, predictions)
    ])
```
